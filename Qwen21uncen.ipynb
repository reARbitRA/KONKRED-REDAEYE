### ۲. دانلود فایل مدل GGUF

مدل را از Hugging Face دانلود می‌کنیم. ممکن است این مرحله کمی طول بکشد.

```python
from huggingface_hub import hf_hub_download

model_name = "abenzerps/Qwen-Image-2.1-Uncensored-GGUF"
model_file = "qwen-image-2.1-uncensored-q8_0.gguf" # یا هر نام فایل GGUF دیگری که می‌خواهید

model_path = hf_hub_download(repo_id=model_name, filename=model_file)
print(f"مدل در مسیر: {model_path} دانلود شد.")
```

### ۳. بارگذاری مدل

حالا مدل را با استفاده از `Llama` بارگذاری می‌کنیم. تعداد `n_gpu_layers` را می‌توانید بر اساس GPU موجود تنظیم کنید (برای استفاده از GPU). اگر GPU ندارید یا خطای حافظه می‌گیرید، آن را به ۰ تغییر دهید.

```python
from llama_cpp import Llama

# n_gpu_layers را با توجه به مشخصات GPU خود تنظیم کنید
# اگر GPU ندارید، یا خطای حافظه می‌گیرید، آن را 0 بگذارید.
llm = Llama(
    model_path=model_path,
    n_gpu_layers=-1,  # -1 برای استفاده از تمام لایه‌های GPU در صورت امکان
    n_ctx=4096,       # حداکثر طول متن (context) برای مدل
    n_batch=512,      # اندازه batch برای پردازش
    verbose=False     # نمایش لاگ‌های llama.cpp
)
print("مدل با موفقیت بارگذاری شد.")
```

### ۴. اجرای Inference (مثال)

اکنون می‌توانید از مدل برای تولید پاسخ استفاده کنید.

```python
prompt = "سلام، حالت چطوره؟"

output = llm(
    prompt,
    max_tokens=100,  # حداکثر تعداد توکن برای تولید
    stop=["Qwen-Image", "User:", "\n"], # کلماتی که باعث توقف تولید می‌شوند
    echo=True       # نمایش prompt در خروجی
)

print(output["choices"][0]["text"])
```

### توجه:

*   **`model_file`**: ممکن است نام فایل GGUF در مخزن Hugging Face متفاوت باشد. حتماً آن را از صفحه مدل بررسی و به روز کنید.
*   **`n_gpu_layers`**: تنظیم این پارامتر برای استفاده از GPU بسیار مهم است. اگر با خطاهای مربوط به GPU مواجه شدید، آن را به `0` تغییر دهید تا مدل فقط از CPU استفاده کند.
*   **زمان دانلود**: دانلود مدل‌های بزرگ ممکن است زمان‌بر باشد.
